# Historical Movers Analysis

This notebook analyzes historical Kraken data to identify the most consistent high-volume movers.

```python
import sys
import os
import pandas as pd

# Ensure project root is in path
project_root = os.path.abspath(os.path.join(os.getcwd(), '..', '..', 'src'))
if project_root not in sys.path:
    sys.path.append(project_root)

from ggTrader.data.kraken.historical_data import KrakenHistoricalData
```

```python
k = KrakenHistoricalData()

# Load historical movers data from parquet
df = k.load_historical_movers_from_parquet()

print(df.head())

# Identify top 20 symbols by volume per day
top10_per_day = (
    df.sort_values(["date", "volume"], ascending=[True, False])
    .groupby("date")
    .head(20)
)

# Count frequency of each symbol in daily top-20s
top10_counts = (
    top10_per_day.groupby("symbol")
    .size()
    .sort_values(ascending=False)
    .rename("top10_days_count")
    .reset_index()
)

print("\nMost frequent daily top movers:")
print(top10_counts.head(20))
```
